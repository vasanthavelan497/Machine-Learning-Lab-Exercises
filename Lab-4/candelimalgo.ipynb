```python
import csv
data = []
with open("finds.csv", "r") as csvfile:
    reader = csv.reader(csvfile)
    for row in reader:
        data.append(row)
num_attr = len(data[0]) - 1
S = ['0'] * num_attr
G = [['?' for i in range(num_attr)] for j in range(num_attr)]

print("\nInitial Specific Hypothesis (S):")
print(S)
print("\nInitial General Hypothesis (G):")
for g in G:
    print(g)

for i in range(len(data)):
    # Positive Example
    if data[i][num_attr] == "Yes":
        if S == ['0'] * num_attr:
            S = data[i][:num_attr]

        for j in range(num_attr):
            if S[j] != data[i][j]:
                S[j] = '?'
                G[j][j] = '?'

    # Negative Example
    else:
        for j in range(num_attr):
            if S[j] != data[i][j]:
                G[j][j] = S[j]
            else:
                G[j][j] = '?'

    print("\nTraining Instance", i + 1)
    print("Specific Hypothesis (S):", S)
    print("General Hypothesis (G):")
    for g in G:
        print(g)
GH = []
for g in G:
    if g != ['?'] * num_attr:
        GH.append(g)
print("\nFinal Specific Hypothesis (S):")
print(S)
print("\nFinal General Hypothesis (G):")
for g in GH:
    print(g)
```
