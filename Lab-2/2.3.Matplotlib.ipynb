```python
import matplotlib.pyplot as plt
import numpy as np
import math
x=np.arange(0,math.pi*2,0.1)
y=np.sin(x)
z=np.cos(x)
a=np.tan(x)
plt.plot(x,y)
plt.plot(x,z)
plt.plot(x,a)
```

```python
from matplotlib import pyplot as plt
import numpy as np
import math
x = np.linspace(-np.pi,np.pi,100)
y=np.sin(x)
plt.plot(x,y)
plt.xlabel('angle')
plt.ylabel('sine function')
plt.title('sine function')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])

fig, a = plt.subplots(2, 2)

# First graph (top-left)
a[0][0].plot(x, x*x)
a[0][0].set_title("x²")

# Second graph (bottom-right)
a[1][1].plot(x, x*x*x)
a[1][1].set_title("x³")

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,20,30,40,50]

plt.plot(x, y)
plt.grid(True)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.plot(x,y,linestyle='--')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1,2,3,4,5])

fig, a = plt.subplots(2,2)

a[0][0].plot(x, x*x)
plt.grid(True)

a[0][0].set_title("Square")

a[0][1].plot(x,x)
a[0][1].set_title("Linear")

a[1][0].plot(x, x**3)
a[1][0].set_title("Cube")

a[1][1].plot(x, x+5)
a[1][1].set_title("x+5")

plt.show()
```
