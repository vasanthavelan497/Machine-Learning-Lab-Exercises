```python
import pandas as pd
data = {
    'marks1': pd.Series([37, 40, 42, 57, 60], index=[100, 101, 102, 103, 104]),
    'marks2': pd.Series([40, 42, 43, 58, 67], index=[100, 101, 102, 103, 104])
}
df = pd.DataFrame.from_dict(data)
print(df['marks1'])
df['total'] = df['marks1'] + df['marks2']
print(df)
                          
```
