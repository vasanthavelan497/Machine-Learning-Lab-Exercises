```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Iris dataset
df1 = sns.load_dataset("iris")

# Histogram
sns.histplot(df1['sepal_width'])
plt.grid(True)
plt.title("Distribution plot for sepal_width")
plt.show()

# KDE plot
sns.kdeplot(df1['sepal_width'])
plt.grid(True)
plt.title("Distribution plot for sepal_width")
plt.show()
```
